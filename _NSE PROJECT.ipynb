```python
# -------------------------------
# Inventory Management System
# Python Mini Project
# -------------------------------

# List to store inventory records (10 items)
inventory = [
    {"ID": 1, "Name": "Laptop", "Category": "ELECTRONICS", "Quantity": 10, "Price": 55000},
    {"ID": 2, "Name": "Mouse", "Category": "ELECTRONICS", "Quantity": 50, "Price": 500},
    {"ID": 3, "Name": "Keyboard", "Category": "ELECTRONICS", "Quantity": 30, "Price": 1200},
    {"ID": 4, "Name": "Monitor", "Category": "ELECTRONICS", "Quantity": 15, "Price": 12000},
    {"ID": 5, "Name": "Chair", "Category": "FURNITURE", "Quantity": 20, "Price": 2500},
    {"ID": 6, "Name": "Table", "Category": "FURNITURE", "Quantity": 10, "Price": 4500},
    {"ID": 7, "Name": "Notebook", "Category": "STATIONERY", "Quantity": 100, "Price": 50},
    {"ID": 8, "Name": "Pen", "Category": "STATIONERY", "Quantity": 200, "Price": 10},
    {"ID": 9, "Name": "Printer", "Category": "ELECTRONICS", "Quantity": 8, "Price": 15000},
    {"ID": 10, "Name": "Router", "Category": "NETWORKING", "Quantity": 12, "Price": 3000}
]

# Function to add an item
def add_item():
    try:
        item_id = int(input("Enter Item ID: "))
        name = input("Enter Item Name: ").strip().title()
        category = input("Enter Category: ").strip().upper()
        quantity = int(input("Enter Quantity: "))
        price = float(input("Enter Price: "))

        item = {
            "ID": item_id,
            "Name": name,
            "Category": category,
            "Quantity": quantity,
            "Price": price
        }

        inventory.append(item)
        print("Item added successfully!")

    except ValueError:
        print("Invalid input! Please enter correct values.")

# Function to view inventory
def view_inventory():
    if len(inventory) == 0:
        print("No items available in inventory.")
    else:
        print("\n--- Inventory Records ---")
        for item in inventory:
            print("ID:", item.get("ID"))
            print("Name:", item.get("Name"))
            print("Category:", item.get("Category"))
            print("Quantity:", item.get("Quantity"))
            print("Price:", item.get("Price"))
            print("-------------------------")

# Function to search item by category
def search_item():
    search_category = input("Enter category to search: ").strip().upper()
    found = False

    for item in inventory:
        if item.get("Category") == search_category:
            print(item)
            found = True

    if not found:
        print("No items found in this category.")

# Function to save inventory to file
def save_to_file():
    with open("inventory_records.txt", "w") as file:
        for item in inventory:
            file.write(str(item) + "\n")
    print("Inventory records saved successfully!")

# Menu-driven program
while True:
    print("\n===== Inventory Management Menu =====")
    print("1. Add Item")
    print("2. View Inventory")
    print("3. Search Item by Category")
    print("4. Save & Exit")

    choice = input("Enter your choice (1-4): ").strip()

    if choice == "1":
        add_item()
    elif choice == "2":
        view_inventory()
    elif choice == "3":
        search_item()
    elif choice == "4":
        save_to_file()
        print("Exiting program. Thank you!")
        break
    else:
        print("Invalid choice. Please select between 1 and 4.")
```
